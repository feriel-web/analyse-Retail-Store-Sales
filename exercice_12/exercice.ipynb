```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os


titanic = sns.load_dataset("titanic")
taxis = sns.load_dataset("taxis")

# Créer le dossier figures
os.makedirs("figures", exist_ok=True)
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

sns.histplot(data=titanic, x="age", bins=30, ax=ax)

ax.set_title("Répartition des âges des passagers")
ax.set_xlabel("Âge (années)")
ax.set_ylabel("Nombre de passagers")

plt.tight_layout()
fig.savefig("figures/01_age_passagers.png", dpi=150)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(
    data=titanic,
    x="pclass",
    y="survived",
    hue="sex",
    ax=ax
)

ax.set_title("Taux de survie selon la classe et le sexe")
ax.set_xlabel("Classe du billet")
ax.set_ylabel("Taux de survie")
ax.set_ylim(0, 1)

plt.tight_layout()
fig.savefig("figures/02_survie_classe_sexe.png", dpi=150)
plt.show()
```

Le graphique montre la répartition des âges des passagers du Titanic. Il permet d'identifier les tranches d'âge les plus représentées et de voir si les passagers étaient principalement jeunes ou adultes.

Le graphique compare les taux de survie des hommes et des femmes selon leur classe. Il permet de voir si le sexe et la classe étaient associés aux chances de survie

```python
fig, ax = plt.subplots(figsize=(9, 5))

sns.boxplot(
    data=titanic,
    x="pclass",
    y="fare",
    ax=ax
)

ax.set_title("Prix des billets selon la classe")
ax.set_xlabel("Classe du billet")
ax.set_ylabel("Prix du billet")

plt.tight_layout()
fig.savefig("figures/03_prix_classe.png", dpi=150)
plt.show()
```

Le graphique montre la répartition des prix des billets selon la classe. Il permet de comparer les médianes et de constater si les prix sont généralement plus élevés dans les classes supérieures.

```python
taxis["pickup"] = pd.to_datetime(taxis["pickup"])
taxis["jour"] = taxis["pickup"].dt.date

ca_journalier = taxis.groupby("jour")["total"].sum()
fig, ax = plt.subplots(figsize=(12, 5))

ca_journalier.plot(kind="line", ax=ax)

ax.set_title("Chiffre d'affaires des taxis jour après jour")
ax.set_xlabel("Jour")
ax.set_ylabel("Chiffre d'affaires")
ax.tick_params(axis="x", rotation=45)

plt.tight_layout()
fig.savefig("figures/04_ca_journalier.png", dpi=150)
plt.show()
```
