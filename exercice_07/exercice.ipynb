```python
import seaborn as sns
import pandas as pd
titanic = sns.load_dataset("titanic")

mineurs = titanic[titanic["age"] < 18]
nombre_mineurs = len(mineurs)

print("Moins de 18 ans :", nombre_mineurs)


billets_plus_chers = (
    titanic.sort_values("fare", ascending=False)
    [["fare", "class", "embark_town"]]
    .head(5)
)

print(billets_plus_chers)


passagers_1re = titanic[titanic["pclass"] == 1]

femmes_1re = passagers_1re[
    passagers_1re["sex"] == "female"
]

part_femmes_1re = (
    len(femmes_1re) / len(passagers_1re) * 100
)

print("Femmes en 1re classe :", round(part_femmes_1re, 1), "%")

taux_survie_femmes_1re = femmes_1re["survived"].mean() * 100

print(
    "Survie femmes 1re :",
    round(taux_survie_femmes_1re, 1),
    "%"
)


hommes_3e = titanic[
    (titanic["sex"] == "male") &
    (titanic["pclass"] == 3)
]

taux_survie_hommes_3e = hommes_3e["survived"].mean() * 100

print(
    "Survie hommes 3e :",
    round(taux_survie_hommes_3e, 1),
    "%"
)
titanic.duplicated().sum()
```
