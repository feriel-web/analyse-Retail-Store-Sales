```python

import pandas as pd

# Chargement des données
df = pd.read_csv("../data/raw/retail_store_sales.csv")

# Inspection générale
print("Dimensions du dataset :", df.shape)

# Informations sur les colonnes
df.info()

# Vérification des valeurs manquantes
print("\nValeurs manquantes :")
print(df.isna().sum())

# Vérification des doublons
print("\nNombre de doublons :")
print(df.duplicated().sum())

# Statistiques descriptives
display(df.describe(include="all"))
```

```python
# Analyse de la répartition des valeurs manquantes

df["Nb valeurs manquantes"] = df.isna().sum(axis=1)

print("Nombre de lignes selon le nombre de valeurs manquantes :")
print(df["Nb valeurs manquantes"].value_counts().sort_index())

print("\nExemples de lignes contenant des valeurs manquantes :")
display(df[df["Nb valeurs manquantes"] > 0].head(10))
```

```python
# Analyse des valeurs manquantes par colonne

colonnes = [
    "Item",
    "Price Per Unit",
    "Quantity",
    "Total Spent",
    "Discount Applied"
]

print("Combinaisons de valeurs manquantes :")
display(
    df[colonnes].isna().value_counts().reset_index(name="Nombre de lignes")
)

# Vérifier les lignes où les trois données financières sont absentes
lignes_sans_donnees_financieres = df[
    df[["Price Per Unit", "Quantity", "Total Spent"]].isna().all(axis=1)
]

print(
    "\nLignes où les trois données financières sont absentes :",
    len(lignes_sans_donnees_financieres)
)

display(lignes_sans_donnees_financieres.head(10))
```

```python
# Vérification des données financières

# Cas 1 : le prix manque, mais le montant et la quantité existent
cas_prix_manquant = df[
    df["Price Per Unit"].isna()
    & df["Total Spent"].notna()
    & df["Quantity"].notna()
]

print("Prix manquant mais calculable :", len(cas_prix_manquant))

# Cas 2 : la quantité manque, mais le prix et le montant existent
cas_quantite_manquante = df[
    df["Quantity"].isna()
    & df["Price Per Unit"].notna()
    & df["Total Spent"].notna()
]

print("Quantité manquante mais calculable :", len(cas_quantite_manquante))

# Cas 3 : le montant manque, mais le prix et la quantité existent
cas_total_manquant = df[
    df["Total Spent"].isna()
    & df["Price Per Unit"].notna()
    & df["Quantity"].notna()
]

print("Montant manquant mais calculable :", len(cas_total_manquant))

# Afficher quelques exemples
display(cas_prix_manquant.head())
display(cas_quantite_manquante.head())
display(cas_total_manquant.head())
```

```python
# Identifier les prix unitaires manquants
# alors que le montant total et la quantité sont disponibles

cas_prix_reconstituable = df[
    df["Price Per Unit"].isna()
    & df["Total Spent"].notna()
    & df["Quantity"].notna()
].copy()

# Calcul du prix unitaire
cas_prix_reconstituable["Prix calculé"] = (
    cas_prix_reconstituable["Total Spent"]
    / cas_prix_reconstituable["Quantity"]
)

print("Nombre de prix calculables :", len(cas_prix_reconstituable))

display(
    cas_prix_reconstituable[
        ["Category", "Item", "Quantity", "Total Spent", "Prix calculé"]
    ].head(15)
)

print("\nNombre de prix calculés différents :")
print(cas_prix_reconstituable["Prix calculé"].nunique())

print("\nStatistiques des prix calculés :")
display(cas_prix_reconstituable["Prix calculé"].describe())
```

```python
# Comparer les prix calculés aux prix déjà présents dans le dataset

prix_existants = set(df["Price Per Unit"].dropna().unique())

prix_calcules = set(cas_prix_reconstituable["Prix calculé"].unique())

prix_non_repertories = prix_calcules - prix_existants

print("Nombre de prix calculés :", len(prix_calcules))
print("Nombre de prix déjà présents :", len(prix_existants))
print("Prix calculés absents des prix existants :", prix_non_repertories)

# Vérifier si les prix calculés sont cohérents avec les prix existants
print(
    "\nTous les prix calculés existent-ils déjà dans le dataset ?",
    prix_non_repertories == set()
)
```

```python
# Correction des prix unitaires manquants

# Compter les valeurs manquantes avant correction
manquants_avant = df["Price Per Unit"].isna().sum()

# Reconstituer les prix unitaires
masque = (
    df["Price Per Unit"].isna()
    & df["Total Spent"].notna()
    & df["Quantity"].notna()
)

df.loc[masque, "Price Per Unit"] = (
    df.loc[masque, "Total Spent"]
    / df.loc[masque, "Quantity"]
)

# Compter les valeurs manquantes après correction
manquants_apres = df["Price Per Unit"].isna().sum()

print("Prix manquants avant correction :", manquants_avant)
print("Prix reconstitués :", masque.sum())
print("Prix manquants après correction :", manquants_apres)
```

**Traitement des valeurs manquantes – Price Per Unit**

La colonne `Price Per Unit` contenait initialement 609 valeurs manquantes.

Après analyse, nous avons constaté que ces lignes disposaient toutes d'un montant total (`Total Spent`) et d'une quantité (`Quantity`) renseignés.

Nous avons donc pu reconstituer les 609 prix unitaires grâce à la formule :

\[
Price\ Per\ Unit = \frac{Total\ Spent}{Quantity}
\]

Une vérification a confirmé que les 25 prix différents obtenus existaient déjà dans les valeurs observées de cette colonne.

Cette méthode permet de récupérer les informations manquantes sans supprimer de transactions ni remplacer les valeurs par une moyenne arbitraire.

```python
# Vérifier si les quantités manquantes peuvent être reconstituées

masque_quantite = (
    df["Quantity"].isna()
    & df["Price Per Unit"].notna()
    & df["Total Spent"].notna()
)

print("Quantités manquantes :", df["Quantity"].isna().sum())
print("Quantités reconstituables :", masque_quantite.sum())
```

**Traitement des valeurs manquantes – Quantity**

La colonne `Quantity` contenait initialement 604 valeurs manquantes.

Nous avons vérifié si ces quantités pouvaient être reconstituées à partir du prix unitaire (`Price Per Unit`) et du montant total (`Total Spent`).

Cependant, aucune des 604 lignes ne disposait à la fois d'un prix unitaire et d'un montant total renseignés. Il était donc impossible de retrouver les quantités manquantes de manière fiable.

Nous avons choisi de conserver ces valeurs manquantes plutôt que de les remplacer par une moyenne ou une valeur arbitraire, afin de ne pas introduire de données artificielles.

```python
print("Montants manquants :", df["Total Spent"].isna().sum())

masque_total = (
    df["Total Spent"].isna()
    & df["Price Per Unit"].notna()
    & df["Quantity"].notna()
)

print("Montants reconstituables :", masque_total.sum())
```

**Traitement des valeurs manquantes – Total Spent**

La colonne `Total Spent` contenait initialement 604 valeurs manquantes.

Nous avons vérifié s'il était possible de recalculer ces montants à partir du prix unitaire (`Price Per Unit`) et de la quantité (`Quantity`).

Cependant, aucune des 604 lignes ne disposait simultanément de ces deux informations. Le montant total ne pouvait donc pas être reconstitué de manière fiable.

Nous avons choisi de conserver les valeurs manquantes afin de ne pas inventer de données ou fausser les résultats de notre analyse.

```python
# Examiner les valeurs manquantes dans la colonne Item

print("Valeurs manquantes dans Item :", df["Item"].isna().sum())

print("\nRépartition des valeurs manquantes par catégorie :")
display(
    df[df["Item"].isna()]["Category"].value_counts()
)
```

**Traitement des valeurs manquantes – Item**

La colonne `Item` contenait initialement 1 213 valeurs manquantes.

Nous avons examiné leur répartition selon la colonne `Category`. Les valeurs manquantes sont réparties dans les huit catégories du jeu de données.

Même si la catégorie de chaque transaction est connue, elle ne permet pas de retrouver précisément le nom de l'article acheté.

Nous avons donc choisi de conserver ces valeurs manquantes plutôt que d'attribuer des noms d'articles au hasard, afin de préserver la fiabilité des données.

```python
# Examiner les valeurs de la colonne Discount Applied

print("Valeurs manquantes :", df["Discount Applied"].isna().sum())

print("\nRépartition des valeurs renseignées :")
display(df["Discount Applied"].value_counts(dropna=False))
```

**Traitement des valeurs manquantes – Discount Applied**

La colonne `Discount Applied` contenait initialement 4 199 valeurs manquantes.

Les valeurs renseignées indiquaient 4 219 transactions avec une réduction (`True`) et 4 157 transactions sans réduction (`False`).

Nous ne disposons pas d'informations permettant de déterminer si les valeurs manquantes correspondent à des transactions avec ou sans réduction.

Nous avons donc choisi de conserver ces valeurs manquantes plutôt que de les remplacer automatiquement par `False`, afin de ne pas fausser les résultats de l'analyse.

```python
# Vérification des types de données

print("Types de données des colonnes :")
display(df.dtypes)

print("\nInformations générales du DataFrame :")
df.info()
```

```python
# Correction des types de données

# Convertir Transaction Date en date
df["Transaction Date"] = pd.to_datetime(
    df["Transaction Date"],
    errors="coerce"
)

# Convertir Quantity en entier tout en conservant les valeurs manquantes
df["Quantity"] = df["Quantity"].astype("Int64")

# Convertir Discount Applied en booléen tout en conservant les valeurs manquantes
df["Discount Applied"] = df["Discount Applied"].astype("boolean")


# Vérifier les nouveaux types
print("Nouveaux types de données :")
display(df.dtypes)

print("\nNombre de dates non converties :",
      df["Transaction Date"].isna().sum())
```

**Vérification et correction des types de données**

Après vérification des types de données, nous avons identifié trois colonnes nécessitant une conversion.

- `Transaction Date` : convertie au format datetime afin de faciliter les analyses temporelles.
- `Quantity` : convertie au format entier nullable (`Int64`) afin de représenter des quantités entières tout en conservant les valeurs manquantes.
- `Discount Applied` : convertie au format booléen nullable (`boolean`) afin de représenter correctement les valeurs True, False et les valeurs manquantes.

```python
# Détection et analyse des valeurs aberrantes

colonnes_numeriques = [
    "Price Per Unit",
    "Quantity",
    "Total Spent"
]

display(df[colonnes_numeriques].describe())
```

```python
# Détection des valeurs aberrantes avec la méthode IQR

colonnes = ["Price Per Unit", "Quantity", "Total Spent"]

for colonne in colonnes:
    Q1 = df[colonne].quantile(0.25)
    Q3 = df[colonne].quantile(0.75)

    IQR = Q3 - Q1

    limite_basse = Q1 - 1.5 * IQR
    limite_haute = Q3 + 1.5 * IQR

    aberrantes = df[
        (df[colonne] < limite_basse)
        | (df[colonne] > limite_haute)
    ]

    print(f"\nColonne : {colonne}")
    print("Limite basse :", limite_basse)
    print("Limite haute :", limite_haute)
    print("Nombre de valeurs potentiellement aberrantes :", len(aberrantes))
```

```python
# Examiner les montants potentiellement aberrants

Q1 = df["Total Spent"].quantile(0.25)
Q3 = df["Total Spent"].quantile(0.75)
IQR = Q3 - Q1

limite_haute = Q3 + 1.5 * IQR

valeurs_aberrantes = df[
    df["Total Spent"] > limite_haute
]

display(
    valeurs_aberrantes[
        ["Transaction ID", "Category", "Item",
         "Price Per Unit", "Quantity", "Total Spent"]
    ].sort_values("Total Spent", ascending=False)
)
```

**Bilan des valeurs aberrantes**

L'analyse a révélé 60 valeurs potentiellement aberrantes dans `Total Spent`. Après vérification, elles sont cohérentes : 41 × 10 = 410. Nous avons donc choisi de les conserver. Aucune valeur aberrante n'a été détectée dans `Price Per Unit` et `Quantity`.
