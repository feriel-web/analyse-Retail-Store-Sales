```python
import pandas as pd
import os

df = pd.read_csv("../exercice_08/data/titanic_propre.csv")

df.groupby(["pclass", "sex"])["survived"].mean().round(2)
```

Les femmes ont globalement un taux de survie supérieur à celui des hommes dans chaque classe. On peut donc conclure que le sexe a joué un rôle important dans les chances de survie

```python
df["tranche_age"] = pd.cut(
    df["age"],
    bins=[0, 12, 18, 60, 100],
    labels=["enfant", "ado", "adulte", "senior"],
    include_lowest=True
)

df.groupby("tranche_age", observed=False)["survived"].mean().round(2)
```

Les enfants présentent un taux de survie de 58% contre 37% pour les adultes. Cela permet de voir si les enfants ont bénéficié d'une priorité lors de l'évacuation

```python
df.groupby("embark_town")["fare"].mean().round(2).sort_values(ascending=False)
```

```python
pd.crosstab(
    df["embark_town"],
    df["pclass"],
    normalize="index"
).round(2)
```

Cherbourg accueille la plus grande proportion de passagers de première classe (51 %), tandis que Queenstown accueille principalement des passagers de troisième classe (94 %).

Cela peut expliquer pourquoi le prix moyen des billets est différent selon le port : les passagers de première classe payaient généralement plus cher
