```python
import pandas as pd
import numpy as np
import seaborn as sns
import os

# Charger les données Titanic
df = sns.load_dataset("titanic")

# Afficher les premières lignes
df.head()

df.shape

df.info()
```

```python
df.isna().sum()

df.isna().mean().round(3)

mediane_age = df.groupby(
    ["pclass", "sex"]
)["age"].transform("median")

df["age"] = df["age"].fillna(mediane_age)

df["age"].isna().sum()
```

```python
df["embark_town"] = df["embark_town"].fillna("Inconnu")

df["embark_town"].isna().sum()
```

```python
df = df.drop(columns=["deck"])
```

```python

# Compter les doublons
print("Nombre de doublons :", df.duplicated().sum())

# Afficher les doublons
display(df[df.duplicated(keep=False)])

# on conserve les doublons,
# car deux passagers peuvent avoir les mêmes caractéristiques


# 4. Supprimer les colonnes redondantes

df = df.drop(columns=["alive", "class"])

# Corriger les types des colonnes
df["sex"] = df["sex"].astype("category")
df["pclass"] = df["pclass"].astype("category")
df["embarked"] = df["embarked"].astype("category")
df["embark_town"] = df["embark_town"].astype("category")
df["survived"] = df["survived"].astype(bool)

# Vérifier les types
print("\nTypes des colonnes :")
print(df.dtypes)




# Calculer les quartiles
q1 = df["fare"].quantile(0.25)
q3 = df["fare"].quantile(0.75)

# Calculer l'écart interquartile
iqr = q3 - q1

# Calculer les bornes
borne_basse = q1 - 1.5 * iqr
borne_haute = q3 + 1.5 * iqr

print("\nBorne basse :", borne_basse)
print("Borne haute :", borne_haute)

# Identifier les billets hors bornes
billets_hors_bornes = df[
    (df["fare"] < borne_basse) |
    (df["fare"] > borne_haute)
]

# Afficher les billets concernés
display(billets_hors_bornes[["fare", "pclass", "sex"]])

# Compter les billets hors bornes
print("Nombre de billets hors bornes :", len(billets_hors_bornes))

# on conserve ces billets,
# car un prix élevé n'est pas forcément une erreur
```

```python
# 6. Enregistrer le fichier nettoyé

import os

# Créer le dossier data s'il n'existe pas
os.makedirs("data", exist_ok=True)

# Enregistrer le tableau
df.to_csv("data/titanic_propre.csv", index=False)

print("\nFichier enregistré avec succès !")


print("\nDimensions du tableau :", df.shape)

print("\nValeurs manquantes restantes :")
print(df.isna().sum())

print("\nNombre de doublons :", df.duplicated().sum())

print("\nAperçu du tableau nettoyé :")
display(df.head())
```
