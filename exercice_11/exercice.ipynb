```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

taxis = sns.load_dataset("taxis")
taxis.shape
taxis.head()
taxis.info()
```

```python
taxis["pickup"] = pd.to_datetime(taxis["pickup"])
taxis["dropoff"] = pd.to_datetime(taxis["dropoff"])
```

```python
taxis["jour"] = taxis["pickup"].dt.date

par_jour = taxis.groupby("jour").agg(
    nombre_courses=("pickup", "count"),
    chiffre_affaires=("total", "sum")
)

par_jour
```

```python
taxis["heure"] = taxis["pickup"].dt.hour
courses_par_heure = taxis["heure"].value_counts().sort_index()

courses_par_heure

heure_plus_chargee = courses_par_heure.idxmax()

print("Heure la plus chargée :", heure_plus_chargee)
```

```python
taxis["jour_semaine"] = taxis["pickup"].dt.day_name()
pourboire_moyen = taxis.groupby("jour_semaine")["tip"].mean().round(2)

pourboire_moyen

print("Jour avec le pourboire moyen le plus élevé :")
print(pourboire_moyen.idxmax())
```

```python
taxis["duree_minutes"] = (
    taxis["dropoff"] - taxis["pickup"]
).dt.total_seconds() / 60
```

```python
taxis["duree_minutes"].describe()
q1 = taxis["duree_minutes"].quantile(0.25)
q3 = taxis["duree_minutes"].quantile(0.75)

print("Q1 :", q1)
print("Q3 :", q3)

iqr = q3 - q1

print("IQR :", iqr)
borne_basse = q1 - 1.5 * iqr
borne_haute = q3 + 1.5 * iqr

print("Borne basse :", borne_basse)
print("Borne haute :", borne_haute)

durees_aberrantes = taxis[
    (taxis["duree_minutes"] < borne_basse) |
    (taxis["duree_minutes"] > borne_haute)
]

durees_aberrantes
```

```python
taxis.loc[
    (taxis["duree_minutes"] < 0) |
    (taxis["duree_minutes"] > 180),
    "duree_minutes"
] = pd.NA  
```

Pourquoi utiliser pd.NA ?

Cela signifie que la durée est considérée comme manquante. On conserve les autres informations de la course, comme son montant et son heure de départ
